```python
before = len(df)

df = df.drop_duplicates()

after = len(df)

print("Duplicate rows removed:", before - after)
```

```python
print(df.isnull().sum())
```

```python
# Fill missing numerical values with median
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Fill missing text values with Unknown
text_cols = df.select_dtypes(include="object").columns

for col in text_cols:
    df[col] = df[col].fillna("Unknown")

print("Missing values handled successfully!")
```

```python
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()

print("Text formatting corrected!")
```

```python
date_columns = ["Order Date", "Ship Date"]

for col in date_columns:
    if col in df.columns:
        df[col] = pd.to_datetime(
            df[col],
            errors="coerce"
        )

print("Date conversion completed!")
```

```python
print("Remaining missing values:")
print(df.isnull().sum())

print("\nRemaining duplicates:")
print(df.duplicated().sum())

print("\nCleaned dataset:")
display(df.head())
```

```python
sales_summary = pd.DataFrame({
    "Metric": [
        "Total Sales",
        "Total Profit",
        "Total Orders",
        "Total Customers",
        "Total Products"
    ],
    "Value": [
        df["Sales"].sum(),
        df["Profit"].sum(),
        df["Order ID"].nunique(),
        df["Customer ID"].nunique(),
        df["Product Name"].nunique()
    ]
})

category_summary = (
    df.groupby("Category")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum")
    )
    .reset_index()
)

region_summary = (
    df.groupby("Region")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum")
    )
    .reset_index()
)

top_10_products = (
    df.groupby("Product Name")["Sales"]
    .sum()
    .nlargest(10)
    .reset_index()
)
```

```python
with pd.ExcelWriter(
    "Automated_Data_Report.xlsx",
    engine="openpyxl"
) as writer:

    df.to_excel(
        writer,
        sheet_name="Cleaned Data",
        index=False
    )

    sales_summary.to_excel(
        writer,
        sheet_name="Sales Summary",
        index=False
    )

    category_summary.to_excel(
        writer,
        sheet_name="Category Summary",
        index=False
    )

    region_summary.to_excel(
        writer,
        sheet_name="Region Summary",
        index=False
    )

    top_10_products.to_excel(
        writer,
        sheet_name="Top Products",
        index=False
    )

print("Automated Excel report generated!")
```

```python
def automate_data_report(file_path):

    data = pd.read_csv(file_path)

    # Remove duplicates
    data = data.drop_duplicates()

    # Handle missing numerical values
    for col in data.select_dtypes(include=np.number).columns:
        data[col] = data[col].fillna(data[col].median())

    # Handle missing text values
    for col in data.select_dtypes(include="object").columns:
        data[col] = data[col].fillna("Unknown")
        data[col] = data[col].str.strip()

    # Convert dates
    for col in ["Order Date", "Ship Date"]:
        if col in data.columns:
            data[col] = pd.to_datetime(
                data[col], errors="coerce"
            )

    # Generate summary
    summary = pd.DataFrame({
        "Metric": ["Total Sales", "Total Profit", "Total Orders"],
        "Value": [
            data["Sales"].sum(),
            data["Profit"].sum(),
            data["Order ID"].nunique()
        ]
    })

    # Export reports
    data.to_csv("Cleaned_Data.csv", index=False)

    summary.to_excel(
        "Automated_Summary.xlsx",
        index=False
    )

    print("Automation completed successfully!")

    return data

# Run the automation
cleaned_data = automate_data_report(file_name)
```
